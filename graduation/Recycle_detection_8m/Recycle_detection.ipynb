```python
import os
from ultralytics import YOLO

# --- 설정 ---

# 1. 모델 선택
model = YOLO('yolov8m.pt')

dataset_folder = r'/home/asd/projects/Portfolio/graduation/taco_dataset'


# 2) 위 폴더 경로와 'data.yaml' 파일 이름을 합쳐서 최종 경로를 완성합니다.
data_yaml_path = os.path.join(dataset_folder, 'data.yaml')
```

```python
# 3. 훈련 파라미터 설정
epochs_count = 100
image_size = 640
batch_size = 6
```

```python
# --- 훈련 시작 ---

results = model.train(
    data=data_yaml_path,
    epochs=epochs_count,
    imgsz=image_size,
    batch=batch_size
)
```
